```python
import rasterio
from rasterio.windows import Window
import os
import numpy as np

# Path ke file .tif
input_path = "Google Satellite_2025-05-03_09-53-31.tif"
output_dir = "daerah3_karawang"

# Ukuran patch (misalnya 1024x1024)
patch_size = 1024

# Membuat folder output jika belum ada
os.makedirs(output_dir, exist_ok=True)

with rasterio.open(input_path) as src:
    width = src.width
    height = src.height
    count = 0

    # Loop baris dan kolom berdasarkan ukuran patch
    for i in range(0, height, patch_size):
        for j in range(0, width, patch_size):
            # Cek jika patch berada di luar batas gambar
            if i + patch_size <= height and j + patch_size <= width:
                window = Window(j, i, patch_size, patch_size)
                transform = src.window_transform(window)
                patch = src.read(window=window)

                # Simpan patch ke file baru
                output_path = os.path.join(output_dir, f"patch_{count}.tif")
                with rasterio.open(
                    output_path,
                    'w',
                    driver='GTiff',
                    height=patch.shape[1],
                    width=patch.shape[2],
                    count=src.count,
                    dtype=patch.dtype,
                    crs=src.crs,
                    transform=transform
                ) as dst:
                    dst.write(patch)
                count += 1

print(f"Selesai memotong menjadi {count} patch.")
```

```python
import rasterio
from rasterio.windows import Window
import os
import numpy as np

# Folder input yang berisi file .tif
input_dir = "Tiff"
output_dir = "datasetindo"

# Ukuran patch (misalnya 1024x1024)
patch_size = 1024

# Membuat folder output jika belum ada
os.makedirs(output_dir, exist_ok=True)

# Counter global untuk penamaan patch
global_count = 1

# Loop melalui semua file .tif di folder input
for file_name in os.listdir(input_dir):
    if file_name.endswith(".tif"):
        input_path = os.path.join(input_dir, file_name)

        with rasterio.open(input_path) as src:
            width = src.width
            height = src.height

            # Loop baris dan kolom berdasarkan ukuran patch
            for i in range(0, height, patch_size):
                for j in range(0, width, patch_size):
                    # Cek jika patch berada di luar batas gambar
                    if i + patch_size <= height and j + patch_size <= width:
                        window = Window(j, i, patch_size, patch_size)
                        transform = src.window_transform(window)
                        patch = src.read(window=window)

                        # Simpan patch ke file baru dengan format PNG
                        output_path = os.path.join(output_dir, f"{global_count}.png")
                        with rasterio.open(
                            output_path,
                            'w',
                            driver='PNG',
                            height=patch.shape[1],
                            width=patch.shape[2],
                            count=src.count,
                            dtype=patch.dtype,
                            crs=src.crs,
                            transform=transform
                        ) as dst:
                            dst.write(patch)
                        global_count += 1

print(f"Selesai memotong semua file menjadi {global_count - 1} patch.")
```
